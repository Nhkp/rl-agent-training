# Exercice 1 - Interaction avec CartPole-v1

Objectif : découvrir le cycle observation -> action -> récompense avec l'environnement `CartPole-v1` de Gymnasium.

## Étape 1 - Explorer l'environnement

On crée l'environnement, puis on inspecte l'espace d'observation et l'espace d'action.

```python
import gymnasium as gym

env = gym.make("CartPole-v1")

print("Espace d'observation :", env.observation_space)
print("Type de l'espace d'observation :", type(env.observation_space))
print("Exemple d'observation :", env.observation_space.sample())

print("\nEspace d'action :", env.action_space)
print("Type de l'espace d'action :", type(env.action_space))
print("Exemple d'action :", env.action_space.sample())
```

## Étape 2 - Faire agir un agent aléatoire

L'agent choisit une action au hasard à chaque étape. On lance 10 épisodes complets et on affiche la récompense totale obtenue à chaque épisode.

```python
number_of_episodes = 10

for episode in range(1, number_of_episodes + 1):
    observation, info = env.reset()
    total_reward = 0
    terminated = False
    truncated = False

    while not (terminated or truncated):
        action = env.action_space.sample()
        observation, reward, terminated, truncated, info = env.step(action)
        total_reward += reward

    print(f"Épisode {episode} - récompense totale : {total_reward}")

env.close()
```
